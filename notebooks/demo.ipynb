```python
from PIL import Image
from src.inference import predict_with_heatmap
from src.visualization import overlay_heatmap
import matplotlib.pyplot as plt

im = Image.new('RGB', (224,224), 'gray')
out = predict_with_heatmap(im)
print(out['label'], f"{out['confidence']:.2%}")
if out['heatmap'] is not None:
    overlay = overlay_heatmap(im, out['heatmap'])
    plt.figure(figsize=(6,3))
    plt.subplot(1,2,1); plt.imshow(im); plt.title('original'); plt.axis('off')
    plt.subplot(1,2,2); plt.imshow(overlay); plt.title('Grad-CAM'); plt.axis('off')
    plt.show()
else:
    print('heatmap not available')
```
